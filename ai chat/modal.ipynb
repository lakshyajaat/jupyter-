```python
from transformers import pipeline

qa_model = pipeline("question-answering", model="yikuan8/Clinical-Longformer")

question = "What are the symptoms of asthma?"
context = "Asthma symptoms include coughing, wheezing, shortness of breath, and chest tightness."

result = qa_model(question=question, context=context)
print(result["answer"])
```
