```python
from collections import deque

def bfs(graph, start):
    # Create a queue
    visited = set()
    queue = deque([start])
    visited.add(start)

    # Traverse the graph
    while queue:
        node = queue.popleft()          # Remove the first node in queue
        print(node, end=" ")            # Process (visit) the node

        #Enqueue all unvisited neighbors
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


# Example Graph (Adjacency List)
graph = {
    'A': 
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['C', 'D']
}

# Step 4: Call BFS function
print("BFS Traversal starting from node A:")
bfs(graph, 'A')
```

Recursive

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    print(start, end=" ")  # Process the node
    visited.add(start)

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


# Example Graph (Adjacency List)
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['C', 'D']
}

# Run DFS
print("DFS Traversal starting from node A:")
dfs(graph, 'A')
```

Stack

```python
def dfs_iterative(graph, start):
    visited = set()
    stack = [start]

    while stack:
        node = stack.pop()
        if node not in visited:
            print(node, end=" ")
            visited.add(node)
            # Add neighbors in reverse order (to match recursive order)
            stack.extend(reversed(graph[node]))

# Example Graph
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['C', 'D']
}

print("\nDFS Traversal (Iterative) starting from A:")
dfs_iterative(graph, 'A')
```
