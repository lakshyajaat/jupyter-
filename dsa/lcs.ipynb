```python
# lcs_length_space_opt.py
def lcs_length(A: str, B: str) -> int:
    # ensure B is the shorter string (so we use less memory)
    if len(A) < len(B):
        A, B = B, A
    n, m = len(A), len(B)
    prev = [0] * (m + 1)
    for i in range(1, n + 1):
        cur = [0] * (m + 1)
        ai = A[i - 1]
        for j in range(1, m + 1):
            if ai == B[j - 1]:
                cur[j] = prev[j - 1] + 1
            else:
                cur[j] = prev[j] if prev[j] >= cur[j - 1] else cur[j - 1]
        prev = cur
    return prev[m]

# Example:
if __name__ == "__main__":
    print(lcs_length("AGGTAB", "GXTXAYB"))  # prints 4
```

```python
def lcs_with_sequence(A: str, B: str):
    n, m = len(A), len(B)
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    # fill dp
    for i in range(1, n + 1):
        ai = A[i - 1]
        for j in range(1, m + 1):
            if ai == B[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = dp[i - 1][j] if dp[i - 1][j] >= dp[i][j - 1] else dp[i][j - 1]

    # reconstruct LCS from dp table
    i, j = n, m
    lcs_chars = []
    while i > 0 and j > 0:
        if A[i - 1] == B[j - 1]:
            lcs_chars.append(A[i - 1])
            i -= 1
            j -= 1
        else:
            if dp[i - 1][j] >= dp[i][j - 1]:
                i -= 1
            else:
                j -= 1

    lcs_sequence = ''.join(reversed(lcs_chars))
    return dp[n][m], lcs_sequence

# Example usage:
if __name__ == "__main__":
    A = "AGGTAB"
    B = "GXTXAYB"
    length, seq = lcs_with_sequence(A, B)
    print("LCS length:", length)
    print("LCS sequence:", seq)
```
