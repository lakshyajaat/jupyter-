```python
def naive_string_match(text, pattern):
    n = len(text)
    m = len(pattern)
    positions = []

    for i in range(n - m + 1):   # slide pattern over text
        if text[i:i+m] == pattern: #campair pattern with text
            positions.append(i) # if match found store the location

    return positions


text = input("Enter the text: ")
pattern = input("Enter the pattern to search: ")

matches = naive_string_match(text, pattern)

if matches:
    print("Pattern found at positions:", matches)
else:
    print("Pattern not found")
```

```python
# Function to build LPS array
def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length = 0  # length of previous longest prefix suffix
    i = 1

    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]  # try shorter prefix
            else:
                lps[i] = 0
                i += 1
    return lps


# KMP Search
def kmp_search(text, pattern):
    n, m = len(text), len(pattern)
    lps = compute_lps(pattern)
    positions = []

    i = j = 0  # i -> text index, j -> pattern index

    while i < n:
        if text[i] == pattern[j]:
            i += 1
            j += 1

        if j == m:  # full pattern matched
            positions.append(i - j)
            j = lps[j - 1]  # continue searching

        elif i < n and text[i] != pattern[j]:
            if j != 0:
                j = lps[j - 1]  # use lps to skip
            else:
                i += 1
    return positions


# --- Main Program ---
text = input("Enter the text: ")
pattern = input("Enter the pattern to search: ")

matches = kmp_search(text, pattern)

if matches:
    print("Pattern found at positions:", matches)
else:
    print("Pattern not found")
```

```python
def kmp_search(text, pattern):
    if not pattern:
        return list(range(len(text) + 1))  # handle empty pattern

    # Build  array
    m, n = len(pattern), len(text)
    lps = [0] * m
    j = 0
    for i in range(1, m):
        while j and pattern[i] != pattern[j]:
            j = lps[j-1]
        if pattern[i] == pattern[j]:
            j += 1
            lps[i] = j

    # Search
    res = []
    i = j = 0
    while i < n:
        if text[i] == pattern[j]:
            i += 1
            j += 1
            if j == m:          # full match
                res.append(i - j)
                j = lps[j-1]
        else:
            if j:               # fallback using array
                j = lps[j-1]
            else:               # move forward in text
                i += 1
    return res



text = input("Enter text: ")
pattern = input("Enter pattern: ")
print("Pattern found at positions:", kmp_search(text, pattern) or "Not found")
```
